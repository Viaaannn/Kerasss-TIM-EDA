## Bagian E — Menerapkan tiga tindakan

```python
kolom_hapus = tanda[tanda['tindakan'] == 'hapus']['nama_kolom'].tolist()
aman = df.drop(columns=kolom_hapus).copy()

print(f"Jumlah kolom dibuang: {len(kolom_hapus)}")
print(f"Jumlah kolom tersisa: {aman.shape[1]}")
```

```python
GARAM = 'ganti-dengan-kalimat-rahasia-tim'

def samarkan(nilai, garam=GARAM):
    if pd.isna(nilai):
        return np.nan
    gabungan = garam + str(nilai)
    return hashlib.sha256(gabungan.encode('utf-8')).hexdigest()[:12]

kolom_hash = tanda[tanda['tindakan'] == 'hash']['nama_kolom'].tolist()
for col in kolom_hash:
    aman[col] = aman[col].apply(samarkan)

print(aman[kolom_hash].head(3))
```

```python
# a. Hash dua kali nilai yang sama
val1 = samarkan("PLT-2025-1309")
val2 = samarkan("PLT-2025-1309")
print("Hash identik:", val1 == val2, f"({val1} == {val2})")

# b. Hash dua nilai berbeda
val3 = samarkan("PLT-2025-1512")
print("Hash berbeda:", val1 != val3, f"({val1} != {val3})")
```

**Jawaban Anda:** kalau `GARAM` ikut ter-*commit* ke GitHub, apa yang masih terlindungi dan apa yang tidak?

```python
ACUAN = pd.Timestamp('2025-09-01')

# a. Hitung usia dalam tahun
usia = (ACUAN - df['tanggal_lahir']).dt.days // 365.25

# b. Kelompok umur
batas = [-np.inf, 24, 34, 44, 54, np.inf]
label = ['<25', '25-34', '35-44', '45-54', '55+']
aman['kelompok_umur'] = pd.cut(usia, bins=batas, labels=label)

# c. Buang tanggal_lahir dan kelurahan (jika ada)
aman = aman.drop(columns=['tanggal_lahir', 'kelurahan'], errors='ignore')

# d. Rapikan kecamatan
aman['kecamatan'] = aman['kecamatan'].str.title().str.strip()

print("Kelompok umur terbanyak:", aman['kelompok_umur'].mode()[0])
print("Jumlah kecamatan unik:", aman['kecamatan'].nunique())
```

## Bagian F — Memeriksa sisa risiko

Menghapus nama belum tentu cukup. Untuk setiap kombinasi quasi-identifier, hitung berapa orang yang punya kombinasi sama. Nilai terkecil disebut **k**. Target kelas ini: **k ≥ 5**.

```python
def cek_k(data, kolom):
    grup = data.groupby(kolom).size()
    k_min = grup.min()
    grup_1 = (grup == 1).sum()
    total_grup = len(grup)
    return pd.Series({'k': k_min, 'grup_berisi_1_orang': grup_1, 'total_grup': total_grup})

print(cek_k(df, ['kecamatan', 'jenis_kelamin']))
```

```python
kolom_qi_lama = ['kelurahan', 'jenis_kelamin', 'pendidikan_terakhir', 'tanggal_lahir']
print(cek_k(df, kolom_qi_lama))
```

**Jawaban Anda:** keempat kolom itu tidak memuat nama maupun nomor telepon. Jelaskan dalam dua kalimat mengapa hasilnya tetap berbahaya.

_(Kombinasi dari beberapa quasi-identifier (seperti lokasi kelurahan, jenis kelamin, tingkat pendidikan, dan tanggal lahir secara spesifik) bersifat sangat unik bagi setiap individu. Ketika keempat informasi tersebut digabungkan, mereka secara efektif menjadi penanda unik yang dapat menunjuk langsung ke satu orang tertentu tanpa memerlukan nama.)_

```python
kolom_qi_baru = ['kecamatan', 'jenis_kelamin', 'pendidikan_terakhir', 'kelompok_umur']
print(cek_k(aman, kolom_qi_baru))
```

```python
aman_k5 = aman.copy()

# Opsi: Perlebar kelompok umur dan sederhanakan pendidikan
aman_k5['kelompok_umur_luas'] = pd.cut(
    (ACUAN - df['tanggal_lahir']).dt.days // 365.25,
    bins=[-np.inf, 34, 49, np.inf],
    labels=['<35', '35-49', '50+']
)

# Periksa k-anonymity tanpa kecamatan (atau keluarkan satu QI)
qi_k5 = ['jenis_kelamin', 'pendidikan_terakhir', 'kelompok_umur_luas']
print(cek_k(aman_k5, qi_k5))
```

**Jawaban Anda:** tulis kombinasi yang Anda pilih, nilai k yang dicapai, dan informasi apa yang Anda korbankan untuk mencapainya.
